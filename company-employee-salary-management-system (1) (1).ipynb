```python

# Project: Company Employee Salary Management System

import pandas as pd
import matplotlib.pyplot as plt

# 1. Create sample employee data
data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": [
        "Ali", "Ahmed", "Sara", "Ayesha",
        "Usman", "Fatima", "Hassan", "Zain"
    ],
    "Department": [
        "IT", "HR", "Finance", "IT",
        "Marketing", "HR", "Finance", "IT"
    ],
    "Monthly_Salary": [
        60000, 45000, 55000, 70000,
        40000, 50000, 65000, 48000
    ],
    "Attendance_Days": [26, 24, 25, 27, 22, 26, 25, 24]
}

df = pd.DataFrame(data)

# 2. Display all employee records
print("COMPANY EMPLOYEE MANAGEMENT SYSTEM")
print("=" * 45)
print(df.to_string(index=False))

# 3. Calculate salary statistics
print("\nSALARY REPORT")
print("Total Employees:", len(df))
print("Total Monthly Salary:", df["Monthly_Salary"].sum())
print("Average Salary:", round(df["Monthly_Salary"].mean(), 2))
print("Highest Salary:", df["Monthly_Salary"].max())
print("Lowest Salary:", df["Monthly_Salary"].min())

# 4. Department-wise salary analysis
department_report = df.groupby("Department").agg(
    Employees=("Employee_ID", "count"),
    Total_Salary=("Monthly_Salary", "sum"),
    Average_Salary=("Monthly_Salary", "mean")
).round(2)

print("\nDEPARTMENT-WISE REPORT")
print(department_report)

# 5. Attendance report
print("\nATTENDANCE REPORT")
print(df[["Name", "Attendance_Days"]].to_string(index=False))
print("Average Attendance:", round(df["Attendance_Days"].mean(), 2))

# 6. Salary chart
plt.figure(figsize=(10, 5))
plt.bar(df["Name"], df["Monthly_Salary"])
plt.title("Employee Monthly Salary")
plt.xlabel("Employee Name")
plt.ylabel("Salary")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 7. Department salary chart
department_report["Total_Salary"].plot(
    kind="bar",
    figsize=(8, 5),
    title="Total Salary by Department"
)
plt.xlabel("Department")
plt.ylabel("Total Salary")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# 8. Save reports as CSV files
df.to_csv("employee_records.csv", index=False)
department_report.to_csv("department_salary_report.csv")

print("\nProject completed successfully!")
print("Employee and department reports have been saved.")
```
